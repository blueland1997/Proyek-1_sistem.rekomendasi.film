# Format Dataset

```python
import json

def parse_json_column(df, column_name):
    df[column_name] = df[column_name].apply(lambda x: json.loads(x) if isinstance(x, str) else [])
    df[column_name + '_names'] = df[column_name].apply(lambda x: ', '.join([item['name'] for item in x]) if isinstance(x, list) and x else 'Tidak ada')
    return df
```

```python
json_columns = ['genres', 'keywords', 'production_companies', 'production_countries', 'spoken_languages']
try:
    for col in json_columns:
        if col in df_capstone.columns:
            df_capstone = parse_json_column(df_capstone, col)
    cols_to_drop = [c for c in json_columns if c in df_capstone.columns]
    if cols_to_drop:
        df_capstone = df_capstone.drop(columns=cols_to_drop, errors='ignore')
    rename_map = {col + '_names': col for col in json_columns}
    existing_renames = {k: v for k, v in rename_map.items() if k in df_capstone.columns}
    if existing_renames:
        df_capstone = df_capstone.rename(columns=existing_renames)
except NameError:
    print("Variable df_capstone not found. Run the data-loading cell first.")
```

# Null Dataset

```python
df_capstone.isnull().sum()
```

```python
missing_values_count = df_capstone.isnull().sum()
missing_values_percentage = 100 * df_capstone.isnull().sum() / len(df_capstone)
missing_values_df = pd.DataFrame({
    'Missing Count': missing_values_count,
    'Missing Percentage': missing_values_percentage
})
missing_values_df = missing_values_df[missing_values_df['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)
print("Ringkasan Missing Values:")
display(missing_values_df)
```

```python
try:
    if 'homepage' in df_capstone.columns:
        df_capstone = df_capstone.drop(columns=['homepage'], errors='ignore')
    for col in ['tagline', 'overview']:
        if col in df_capstone.columns:
            df_capstone[col] = df_capstone[col].fillna('Tidak ada')
            df_capstone[col] = df_capstone[col].replace('', 'Tidak ada')

    if 'runtime' in df_capstone.columns:
        median_runtime = df_capstone['runtime'].median(skipna=True)
        df_capstone['runtime'] = df_capstone['runtime'].fillna(median_runtime)
    if 'release_date' in df_capstone.columns:
        mode_vals = df_capstone['release_date'].mode(dropna=True)
        if not mode_vals.empty:
            mode_date = mode_vals.iloc[0]
            df_capstone['release_date'] = df_capstone['release_date'].fillna(mode_date)
        else:
            df_capstone['release_date'] = df_capstone['release_date'].fillna(method='ffill')

    print("homepage dropped (if present).")
    print("Missing values after fill:")
    cols = [c for c in ['tagline','overview','runtime','release_date'] if c in df_capstone.columns]
    print(df_capstone[cols].isnull().sum())
except NameError:
    print("Variable df_capstone not found. Run the data-loading cell first.")
```
